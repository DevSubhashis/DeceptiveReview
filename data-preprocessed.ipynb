```python
# DATA PRE-PROCESSING
```

```python
# We have spacy
# This function performs: Tokenization, Stopword removal, Punctuation removal, Lemmatization

def preprocess_text(text):
    doc = nlp(str(text).lower())
    
    tokens = []
    
    for token in doc:
        # remove stopwords, punctuation, spaces
        if token.is_stop:
            continue
        
        if token.is_punct:
            continue
        
        if token.is_space:
            continue
        
        # keep only alphabetic words
        if not token.is_alpha:
            continue
        
        # lemma
        tokens.append(token.lemma_)
    
    return " ".join(tokens)
```

```python
# Apply Pre-preocessing
```

```python
df["processed_review"] = df["review"].apply(preprocess_text)
```

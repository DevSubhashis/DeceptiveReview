```python
import pandas as pd
import pickle

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("data/processed_reviews.csv")
```

```python
df.sample(2)
```

```python
X = df["processed_review"]
```

```python
y = df["label"]
```

```python
tfidf = TfidfVectorizer(
    
    # Maximum number of features (words)
    max_features=5000,
    
    # Use unigram and bigram
    ngram_range=(1, 2),
    
    # Ignore very common words
    max_df=0.95,
    
    # Ignore very rare words
    min_df=2
)
```

```python
# CONVERT TEXT INTO NUMERICAL FEATURES
```

```python
X_tfidf = tfidf.fit_transform(X)
```

```python
X_tfidf.shape
```

```python
feature_names = tfidf.get_feature_names_out()

feature_names[:20]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    
    X_tfidf,
    y,
    
    # 80% train, 20% test
    test_size=0.2,
    
    # reproducible results
    random_state=42,
    
    # maintain label ratio
    stratify=y
)
```

```python
print("\nTraining Set Shape:")
print(X_train.shape)

print("\nTesting Set Shape:")
print(X_test.shape)

print("\nTraining Labels Shape:")
print(y_train.shape)

print("\nTesting Labels Shape:")
print(y_test.shape)
```

```python
pickle.dump(
    tfidf,
    open("model/tfidf_vectorizer.pkl", "wb")
)

print("\nTF-IDF Vectorizer Saved Successfully!")
```

```python
# SAVE FEATURE DATASET 
```

```python
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\nTF-IDF DataFrame Preview:")

print(tfidf_df.head())

tfidf_df["label"] = y.values

tfidf_df.to_csv(
    "model/tfidf_features.csv",
    index=False
)

print("\nTF-IDF Features Saved Successfully!")
```
